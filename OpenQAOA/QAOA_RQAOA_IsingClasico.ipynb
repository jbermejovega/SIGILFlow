```python
from openqaoa.problems import QUBO

N = 12
J = 1.0
h = 0.5

terms = []
weights = []

# interacción ZZ
for i in range(N - 1):
    terms.append([i, i+1])
    weights.append(-J)

# campo local
for i in range(N):
    terms.append([i])
    weights.append(-h)

qubo = QUBO(N, terms, weights)
```

```python
from openqaoa import QAOA

qaoa = QAOA()
qaoa.set_circuit_properties(p=7) #al aumentar de capas se obtiene mejor resultado

qaoa.compile(qubo)
qaoa.optimize()

result = qaoa.result

print("Energía:", result.optimized["cost"])
print("Solución:", result.most_probable_states)
```

```python
from openqaoa import RQAOA

# Define the RQAOA object (default rqaoa_type = 'adaptive')
r =  RQAOA()

# Set parameters for RQAOA, in this case we fix the n_max to 1 (default), the final cutoff value to 3
r.set_rqaoa_parameters(n_cutoff=3)


r.set_circuit_properties(p=4, param_type='standard', init_type='ramp', mixer_hamiltonian='x')


# Set the classical method used to optimiza over QAOA angles and its properties
r.set_classical_optimizer(method='cobyla', maxiter=200)

r.compile(qubo)
r.optimize()
r.result
```
